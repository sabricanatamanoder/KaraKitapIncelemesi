Makalenin dipnotlarında verilen ki-kare ve Fisher testlerini yeniden üretir.
Sayımlar defterin içinde yazılıdır, veri dosyası gerekmez.

**Çalıştırma:** Çalışma zamanı → Tümünü çalıştır. Birkaç saniye sürer.

```python
# =====================================================================
#  Kara Kitap / Goodreads — dipnotlardaki istatistiklerin denetimi
#  Tek hucre. Veri dosyasi gerektirmez. Colab'da "Calistir" yeterli.
# =====================================================================
from scipy.stats import chi2_contingency, fisher_exact

def ki(ad, a, na, b, nb, iki_yonlu=True):
    """a/na -> b/nb degisimini ki-kare ile sinar."""
    tab = [[a, na - a], [b, nb - b]]
    chi2, p, sd, bek = chi2_contingency(tab)
    pf = fisher_exact(tab)[1]
    kucuk = (bek.min() < 5)
    yildiz = '*' if p < 0.05 else ' '
    ek = f'  (Fisher p = {pf:.3f})' if (kucuk or p < 0.10) else ''
    print(f'{yildiz} {ad:34s} %{100*a/na:5.1f} -> %{100*b/nb:5.1f}   '
          f'chi2 = {chi2:6.2f}  sd = {sd}  p = {p:.3f}{ek}')

def ki3(ad, sayilar, nler):
    """Uc bantli tablo (sd = 2)."""
    tab = [[k, n - k] for k, n in zip(sayilar, nler)]
    chi2, p, sd, _ = chi2_contingency(tab)
    yildiz = '*' if p < 0.05 else ' '
    yuzde = '  '.join(f'%{100*k/n:5.1f}' for k, n in zip(sayilar, nler))
    print(f'{yildiz} {ad:22s} {yuzde}   chi2 = {chi2:6.2f}  sd = {sd}  p = {p:.3f}')

# ---------------------------------------------------------------------
print('DIPNOT 16 — Turkce korpus, D2 (n = 93) -> D3 (n = 161)')
T2, T3 = 93, 161
ki('metin disi otorite',        26, T2,  66, T3)
ki('  Nobel',                    2, T2,  17, T3)
ki('  Pamuk kulliyati',          9, T2,  32, T3)
ki('tema ekseni',               34, T2,  90, T3)
ki('zorluk bildirimi',          23, T2,  44, T3)
ki('uslup sikayeti',            20, T2,  38, T3)
ki('daginiklik sikayeti',       13, T2,  16, T3)
ki('yeniden okuma',             12, T2,  23, T3)
ki('bicim ekseni',              31, T2,  55, T3)

print('\nDIPNOT 17 — Ingilizce korpus, D1 (n = 172) -> D3 (n = 159)')
E1, E3 = 172, 159
ki('yazi ve ustkurmaca',        16, E1,  41, E3)
ki('tasavvuf ve hurufilik',      3, E1,  18, E3)
ki('tema ekseni',               75, E1,  93, E3)
ki('daginiklik sikayeti',       32, E1,  48, E3)
ki('bicim ekseni',              63, E1,  74, E3)
ki('zorlugun odulu',             3, E1,   9, E3)
ki('ceviri farkindaligi',       27, E1,  19, E3)
ki('metin disi otorite',        46, E1,  49, E3)
ki('  Nobel',                    9, E1,   9, E3)
ki('  Pamuk kulliyati',         24, E1,  24, E3)
ki('zorluk bildirimi',          50, E1,  52, E3)

print('\nDIPNOT 14 ve 15 — Tablo 6, puan bantlari (sd = 2)')
TR = [14, 32, 201]
EN = [103, 114, 254]
print('  Turkce   1-2 / 3 / 4-5')
ki3('zorluk ve emek',   [ 4,  9,  50], TR)
ki3('yarida birakma',   [ 1,  1,   7], TR)
ki3('yeniden okuma',    [ 2,  4,  28], TR)
ki3('uslup ve cumle',   [ 7,  9,  37], TR)
ki3('daginiklik',       [ 5,  9,  15], TR)
print('  Ingilizce  1-2 / 3 / 4-5')
ki3('zorluk ve emek',   [26, 37,  76], EN)
ki3('yarida birakma',   [20,  9,   2], EN)
ki3('yeniden okuma',    [ 3,  3,  23], EN)
ki3('uslup ve cumle',   [20, 38,  45], EN)
ki3('daginiklik',       [30, 39,  44], EN)

print('\nYildizli satirlar p < 0,05 duzeyinde anlamlidir.')
```
